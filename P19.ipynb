# Practical 19

**Aim:** Apply data preprocessing techniques: slicing, filtering, sorting, concatenating, aggregating and normalizing data using Pandas.

**Dataset:** [Parking Birmingham](https://archive.ics.uci.edu/dataset/482/parking+birmingham) (UCI Machine Learning Repository) — occupancy readings for 30 NCP car parks in Birmingham, UK, recorded every ~30 minutes between 08:00–16:30 from 2016-10-04 to 2016-12-19.

Columns: `SystemCodeNumber` (car park ID), `Capacity` (max capacity), `Occupancy` (cars parked at that time), `LastUpdated` (timestamp).

### 1. Load Dataset

```python
import pandas as pd

df = pd.read_csv('dataset.csv', parse_dates=['LastUpdated'])
print('Dataset Shape:', df.shape)
df.head(2)
```

### 2. Slicing & Filtering

```python
# Slicing with .iloc and .loc
print('Positional Slicing (.iloc):\n', df.iloc[0:3, 0:3])
print('\nLabel Slicing (.loc):\n', df.loc[0:2, ['SystemCodeNumber', 'Capacity', 'Occupancy']])

# Filtering
filtered = df[(df['Capacity'] > 500) & (df['Occupancy'] >= 400)]
print(f'\nFiltered Count (Capacity > 500 & Occupancy >= 400): {len(filtered)}')
```

### 3. Sorting & Concatenating

```python
# Sorting
print('Sorted by Occupancy (Descending):\n',
      df[['SystemCodeNumber', 'Capacity', 'Occupancy']].sort_values(by='Occupancy', ascending=False).head(3))

# Concatenating (Vertical and Horizontal)
v_concat = pd.concat([df.iloc[:2, :3], df.iloc[2:4, :3]], axis=0)
print('\nVertical Concatenation:\n', v_concat)

h_concat = pd.concat([df[['SystemCodeNumber']].head(3), df[['Occupancy']].head(3)], axis=1)
print('\nHorizontal Concatenation:\n', h_concat)
```

### 4. Aggregating & Normalizing

```python
# Aggregating with groupby
print('Average Occupancy by Car Park:\n', df.groupby('SystemCodeNumber')['Occupancy'].mean().head(4))

# Normalizing Data (Min-Max Scaling)
df['occupancy_norm'] = (df['Occupancy'] - df['Occupancy'].min()) / (df['Occupancy'].max() - df['Occupancy'].min())
print('\nOriginal vs Min-Max Normalized Occupancy:\n', df[['Occupancy', 'occupancy_norm']].head(3))
```
