**Aim:** Use Pandas to create Series and DataFrames, perform data selection, filtering, grouping and aggregation operations.

### 1. Creating Series and DataFrame

```python
import pandas as pd

# Series
salaries = pd.Series([55000, 62000, 78000, 48000], index=["E1", "E2", "E3", "E4"])
print("Salary Series:\n", salaries)

# DataFrame (Employee records)
emp_data = {
    "EmpCode": [201, 202, 203, 204, 205, 206],
    "EmpName": ["Aarav", "Bhavna", "Chirag", "Divya", "Eshan", "Farhan"],
    "Dept": ["Sales", "Tech", "Sales", "HR", "Tech", "HR"],
    "Salary": [58000, 72000, 64000, 51000, 80000, 53000],
    "City": ["Mumbai", "Bengaluru", "Delhi", "Mumbai", "Bengaluru", "Delhi"]
}

df = pd.DataFrame(emp_data)
print("\nEmployee Table:\n", df)
```

### 2. Data Selection

```python
# Columns
print("Names column:\n", df["EmpName"])
print("\nEmpName and Dept:\n", df[["EmpName", "Dept"]])

# iloc
print("\nRows 1 to 3 (iloc):\n", df.iloc[1:4])
print("\nRow 0 details (iloc):\n", df.iloc[0, [1, 2, 3]])

# loc
print("\nRows 0 to 2 for Name and Salary (loc):\n", df.loc[0:2, ["EmpName", "Salary"]])
```

### 3. Data Filtering

```python
# Salary >= 60000
print("High Earner Employees:\n", df[df["Salary"] >= 60000])

# Tech Dept located in Bengaluru
print("\nTech in Bengaluru:\n", df[(df["Dept"] == "Tech") & (df["City"] == "Bengaluru")])
```

### 4. Grouping and Aggregation

```python
# Mean Salary by Dept
print("Average Salary by Dept:\n", df.groupby("Dept")["Salary"].mean())

# Aggregation by City
print("\nCity-wise Salary & Headcount:\n", df.groupby("City").agg({
    "Salary": ["mean", "min", "max"],
    "EmpCode": "count"
}))
```
