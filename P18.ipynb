# Practical 18 (Dataset: Parking Birmingham)

**Aim:** Perform data cleaning operations: detect and handle missing values, remove duplicate records, and treat outliers in a dataset.

**Dataset:** [Parking Birmingham](https://archive.ics.uci.edu/dataset/482/parking+birmingham) (UCI Machine Learning Repository)


### 1. Load Dataset

```python
import pandas as pd
import numpy as np

df = pd.read_csv('dataset.csv')
df.replace('?', np.nan, inplace=True)
print('Initial Shape:', df.shape)
df.head(3)
```

### 2. Detect & Handle Missing Values

```python
# Detect missing values
print('Missing values per column:')
print(df.isnull().sum()[df.isnull().sum() > 0])

# Handle missing values by imputing Median (numeric columns: Capacity, Occupancy)
for col in ['Capacity', 'Occupancy']:
    df[col] = df[col].fillna(df[col].median())

print('\nRemaining missing values:', df.isnull().sum().sum())
```

### 3. Detect & Remove Duplicate Records

```python
# Detect duplicates
print('Duplicate records count:', df.duplicated().sum())

# Remove duplicates
df.drop_duplicates(inplace=True)
print('Shape after removing duplicates:', df.shape)
```

### 4. Detect & Treat Outliers

```python
# Detect outliers using IQR Method on 'Occupancy'
Q1 = df['Occupancy'].quantile(0.25)
Q3 = df['Occupancy'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['Occupancy'] < lower_bound) | (df['Occupancy'] > upper_bound)]
print(f'IQR: {IQR}, Lower Bound: {lower_bound}, Upper Bound: {upper_bound}')
print('Outliers count in Occupancy:', len(outliers))

# Treat outliers using Capping (Winsorization)
df['Occupancy'] = df['Occupancy'].clip(lower=lower_bound, upper=upper_bound)
print('Outliers treated! Capped Min:', df['Occupancy'].min(), 'Capped Max:', df['Occupancy'].max())
```
