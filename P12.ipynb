**Aim:** Use SciPy's stats module to compute statistical measures and perform statistical tests on datasets.

### 1. Statistical Measures using SciPy

```python
from scipy import stats
import numpy as np

# Create Dataset (Daily Temperature in °C)
temps = [18, 21, 22, 22, 25, 27, 28, 30, 31, 35]
print("Dataset:", temps)

# Measures of Central Tendency (Mean, Median, Mode)
print("Mean               :", stats.tmean(temps))
print("Median             :", np.median(temps))
result = stats.mode(temps)
print(f"Mode               : {result.mode} (count: {result.count})")

# Measures of Dispersion & Shape (Std Dev, Variance, Skewness, Kurtosis)
print("Standard Deviation :", round(stats.tstd(temps), 4))
print("Variance           :", round(stats.tvar(temps), 4))
print("Skewness           :", round(stats.skew(temps), 4))
print("Kurtosis           :", round(stats.kurtosis(temps), 4))

# Overall SciPy Describe Summary
print("\nSciPy Describe Summary:\n", stats.describe(temps))
```

### 2. Probability Distributions (Normal Distribution)

```python
# Standard Normal Distribution: Probability Density Function (PDF) & Cumulative Distribution Function (CDF)
x = -0.5
print("Standard Normal PDF at x=-0.5:", round(stats.norm.pdf(x, loc=0, scale=1), 4))
print("Standard Normal CDF at Z=2.0 :", round(stats.norm.cdf(2.0, loc=0, scale=1), 4))
```

### 3. Hypothesis Testing: One-Sample t-Test

```python
# Testing if average temperature matches historical normal (mu=24)
t_stat, p_val = stats.ttest_1samp(temps, popmean=24)

print("Sample:", temps)
print("t-statistic:", round(t_stat, 4))
print("p-value    :", round(p_val, 4))

if p_val < 0.05:
    print("Conclusion : Reject Null Hypothesis (Temperature deviates significantly from 24°C).")
else:
    print("Conclusion : Fail to Reject Null Hypothesis (Temperature conforms to 24°C normal).")
```

### 4. Correlation Test (Pearson Correlation)

```python
# Pearson Correlation Test (Elevation in Meters vs Ambient Temperature)
elevation = [100, 300, 500, 700, 1000]
ambient_temp = [32, 29, 26, 23, 19]

result_corr = stats.pearsonr(elevation, ambient_temp)
print("Correlation : ", round(result_corr.statistic, 4))
print("p-value     : ", round(result_corr.pvalue, 6))
```
